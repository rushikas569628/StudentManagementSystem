# Inheritance, abstraction and polymorphism

```python
class GraduateStudent(Student):
    def __init__(self,student_id,name,age,course,marks,thesis_title):
        super().__init(student_id,name,age,course,marks)
        self.thesis_title=thesis_title

    def display_details(self):
        super().display_details()
        print("Thesis: ",self.thesis_title)

     def calculate_average(self):
         super().calculate_average():
```

```python
class RegularStudent(Student):
     def calculate_average(self):
         super().calculate_average():

     def display_details(self):
        super().display_details()
```

```python
g=GraduateStudent(101, "Anu", 21, " Computer Science ", 90,"AI and its Importance")
g.display_details()
r=RegularStudent(101, "Anu", 21, " ", 19)
```
