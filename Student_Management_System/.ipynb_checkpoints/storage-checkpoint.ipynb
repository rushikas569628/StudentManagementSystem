# Save and load student records using JSON

```python
class Student:
    #class variables
    college_name=""
    student_count=0

    #instance variables
    def __init__(self,student_id,name,age,course,marks):
        self.student_id=student_id
        self.name=name
        self.age=age
        self.course=course
        #protected attribute: only for internal use ->_marks
        #private attribute: 2 underscores-name mangling
        self.__marks=marks

        # class variable student_count shared at class level
        Student.student_count+=1

    def display_details(self):
        print(f"Student_id: {self.student_id}")
        print(f"Name: {self.name}")
        print(f"Age: {self.age}")
        print(f"Course: {self.course}")

    def update_course(self,new_course):
        self.course=new_course
        return new_course

    def calculate_average(self):
        return sum(self.marks)/len(self.marks)

    # decorator tells to treat below method as class method
    #cls=refers class methods
    #self=refers instance methods
    @classmethod
    def get_student_count(cls):
        return cls.student_count

    # staticmethod=A utility function placed inside the class that doesn't automatically receive self or cls
    #we put inside student class for validating student records
    @staticmethod
    def is_valid_mark(mark):
        #isinstance()-builtinfunction whether a value belongs to specified type
        #notinstance()-bool is a subclass of int if it is returns true 
        isinstance(mark,(int,float)) and not isinstance(mark,bool) and 0<=mark<=100

    """
    #create a getter
    def get_marks(self):
        return self.__marks

    #setter-controls what values can be taken
    def set_marks(self,new_marks):
        if isinstance(new_marks,bool):
            raise ValueError("Marks should be between 0 and 100")
        elif not isinstance(new_marks,(int,float)):
            raise ValueError("Marks should be between 0 and 100")
        elif not 0<=new_marks<=100:
            raise ValueError("Marks should be between 0 and 100")

        self.__marks=new_marks
    """
    @property
    def marks(self):
        return self.__marks

    @marks.setter
    def marks(self,new_marks):
        if isinstance(new_marks,bool):
            raise ValueError("Marks must be b/w 0 and 100")

        elif not isinstance(new_marks,(int,float)):
            raise ValueError("Marks must be b/w 0 and 100")
            
        elif not (0<=new_marks<=100):
            raise ValueError("Marks must be b/w 0 and 100")

        self.__marks=new_marks
        
    
```

```python
from abc import ABC,abstractmethod

class StudentStorage(ABC):
    @abstractmethod
    def save_students(self,students):
        pass

    @abstractmethod
    def load_students(self):
        pass
```

```python
import json

class JSONStorage(StudentStorage):
    def __init__(self,filename):
        self.filename=filename
    def save_students(self,students): #students is a list containing 3 objects
        student_data=[]
        for student in students:
            student_dict = {
            "student_id": student.student_id,
            "name": student.name,
            "age": student.age,
            "course": student.course,
            "marks": student.marks
            }
            student_data.append(student_dict)
        with open(self.filename,"w") as file:
            json.dump(student_data,file,indent=4)
            
    def load_students(self):
        print("Loading Students from JSON")
```

```python

student = Student(101, "John", 22, "Computer Science", [85, 90, 88])

storage = JSONStorage("students.json")
storage.save_students([student])
```

Why did we convert the objects into dictionaries?
- Student is a custom Python object.
- JSON doesn't directly support custom Python objects.
- A dictionary containing basic data types can be converted into JSON.
- json.dump(student_data, file, indent=4) writes that data into the file.

```python
class DatabaseStorage(StudentStorage):
    def save_students(self, students):
        print("Saving students to database")

    def load_students(self):
        print("Loading students from database")

storage=[JSONStorage(),DatabaseStorage()]
```

```python
def manage_students(storage):
    storage.save_students([])
    storage.load_students()
```

```python
manage_students(JSONStorage())
manage_students(DatabaseStorage())
```

```python
for s in storage:
    s.save_students([])
    s.load_students()
```
