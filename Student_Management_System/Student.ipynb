# Student class, attributes, constructors and methods

```python
class Student:
    #class variables
    college_name=""
    student_count=0

    #instance variables
    def __init__(self,student_id,name,age,course,marks):
        self.student_id=student_id
        self.name=name
        self.age=age
        self.course=course
        #protected attribute: only for internal use ->_marks
        #private attribute: 2 underscores-name mangling
        self.__marks=marks

        # class variable student_count shared at class level
        Student.student_count+=1

    def display_details(self):
        print(f"Student_id: {self.student_id}")
        print(f"Name: {self.name}")
        print(f"Age: {self.age}")
        print(f"Course: {self.course}")

    def update_course(self,new_course):
        self.course=new_course
        return new_course

    def calculate_average(self):
        return sum(self.marks)/len(self.marks)

    # decorator tells to treat below method as class method
    #cls=refers class methods
    #self=refers instance methods
    @classmethod
    def get_student_count(cls):
        return cls.student_count

    # staticmethod=A utility function placed inside the class that doesn't automatically receive self or cls
    #we put inside student class for validating student records
    @staticmethod
    def is_valid_mark(mark):
        #isinstance()-builtinfunction whether a value belongs to specified type
        #notinstance()-bool is a subclass of int if it is returns true 
        isinstance(mark,(int,float)) and not isinstance(mark,bool) and 0<=mark<=100

    #create a getter
    def get_marks(self):
        return self.__marks

    #setter-controls what values can be taken
    def set_marks(self,new_marks):
        if isinstance(new_marks,bool):
            raise ValueError("Marks should be between 0 and 100")
        elif not isinstance(new_marks,(int,float)):
            raise ValueError("Marks should be between 0 and 100")
        elif not 0<=new_marks<=100:
            raise ValueError("Marks should be between 0 and 100")

        
```
