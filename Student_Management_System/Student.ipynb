```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    @property
    def marks(self):
        return self.__marks

    @marks.setter
    def marks(self, new_marks):
        if 0 <= new_marks <= 100:
            self.__marks = new_marks
        else:
            raise ValueError("Marks must be between 0 and 100.")
```

```python
class Student:
    #class variables
    college_name=""
    student_count=0

    #instance variables
    def __init__(self,student_id,name,age,course,marks):
        self.student_id=student_id
        self.name=name
        self.age=age
        self.course=course
        #protected attribute: only for internal use ->_marks
        #private attribute: 2 underscores-name mangling
        self.marks=marks

        # class variable student_count shared at class level
        Student.student_count+=1

    def display_details(self):
        print(f"Student_id: {self.student_id}")
        print(f"Name: {self.name}")
        print(f"Age: {self.age}")
        print(f"Course: {self.course}")

    def update_age(self,new_age):
        self.age=new_age

    @staticmethod
    def is_valid_age(age):
        return (not isinstance(age,bool) and isinstance(age,int) and 1<=age<=120)

    @property
    def age(self):
        return self.__age

    @age.setter
    def age(self,new_age):
        if not Student.is_valid_age(new_age):
            raise ValueError("age should be b/w 1 to 120")
        self.__age=new_age

    @property
    def course(self):
        return self.__course

    @course.setter
    def course(self,new_course):
        if not isinstance(new_course,str) or not new_course.strip():
            raise ValueError("Course cannot be empty")

        self.__course=new_course.strip()

    @staticmethod
    def is_valid_mark(marks):
           return (
        isinstance(marks, list)
        and len(marks) > 0
        and all(
            not isinstance(mark, bool)
            and isinstance(mark, (int, float))
            and 0 <= mark <= 100
            for mark in marks
        )
    )
    @property
    def marks(self):
        return self.__marks

    @marks.setter
    def marks(self,new_marks):
        if not Student.is_valid_mark(new_marks):
            raise ValueError("marks should be b/w 0 and 100")
        self.__marks=new_marks

    
```
